## Part 3: Selecting, Indexing, and Filtering
* Single brackets `df['Age']` isolate a single structural dimension (**Series**).
* Double brackets `df[['Age']]` encapsulate the selection inside a **DataFrame** format.
* `.loc[]` handles **Label-based** coordinates; `.iloc[]` references strict **Integer-position based** dimensions.
* `.query()` provides readable string logical syntax evaluation.

````python
# 3A: Bracket Selections
series_output = starbucksdrinks['Name']
dataframe_output = starbucksdrinks[['Name', 'Salary']]

print("--- Series Output (single bracket) ---")
display(series_output.head(2))
print("\n--- DataFrame Output (double bracket) ---")
display(dataframe_output.head(2))
````

#### 3B: Deep Index Coordinates via `.loc[]` (Label-based Indexing)

`.loc[]` is used for **label-based** indexing. You provide the row label(s) and column label(s) to select specific data.

````python
print("Value locating via .loc (Row index 2, Label 'Salary'):", starbucksdrinks.loc[2, 'Salary'])
````

#### 3C: Deep Index Coordinates via `.iloc[]` (Integer-position based Indexing)

`.iloc[]` is used for **integer-position based** indexing. You provide the row index(es) and column index(es) (starting from 0) to select data.

````python
print("Value locating via .iloc (Row index 2, Column position 2):", starbucksdrinks.iloc[2, 2])
````

#### 3D: `.query()` Expressions

`.query()` provides a readable way to filter DataFrame rows using a string expression. It's often more intuitive for complex conditions than boolean indexing.

````python
filtered_query = starbucksdrinks.query("Salary > 65000")
print("\n--- Output from .query() Statement (Salary > 65000) ---")
display(filtered_query)
````

### Boolean Mask Filters & Arrays
When nesting multi-conditional query filters across vectors, keep conditional elements securely bound inside their own standalone logical brackets:
* `&` indicates **AND** logical intersection logic.
* `|` indicates **OR** logical union logic.

````python
# Multi-conditional Boolean Mapping Array Masking
mask_condition = (starbucksdrinks['Salary'] >= 70000) & (starbucksdrinks['Age'] <= 32)
display(starbucksdrinks[mask_condition])
````

#### 3E: Utilizing Inclusive Range Bounds with `.between()`

`.between()` is a convenient method for filtering numerical data within a specified inclusive range.

````python
print("\n--- Filter using .between() Method (Salary between 60000 and 80000) ---")
display(starbucksdrinks[starbucksdrinks['Salary'].between(60000, 80000)])
````

#### 3F: List Matching with `.isin()`

`.isin()` is useful for filtering rows where a column's values are present in a given list of values.

````python
print("\n--- Filter using .isin() Method (Name is Alice or Charlie) ---")
display(starbucksdrinks[starbucksdrinks['Name'].isin(['Alice', 'Charlie'])])
````

## Part 5: Aggregation Groupby Routines & Data Merges

````python
left_table = pd.DataFrame({'Emp_ID': [1, 2, 3], 'Role': ['Eng', 'PM', 'Data Sci']})
right_table = pd.DataFrame({'Emp_ID': [2, 3, 4], 'Tenure': [3, 5, 1]})
print("--- Defined Left and Right Tables for Merging ---")
display(left_table)
display(right_table)
````

````python
# 5A: Advanced Groupby & Explicit Named Dictionary Aggregations
mock_regional_data = pd.DataFrame({
    'State': ['CA', 'CA', 'NY', 'NY', 'CA'],
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Count': [295, 239, 150, 110, 340],
    'Year': [2021, 2022, 2021, 2023, 2022]
})

state_summary = mock_regional_data.groupby('State').agg({
    'Count': 'sum',
    'Year': 'max'
})
print("--- Groupby Summary Mapping (.agg) ---")
display(state_summary)
````

#### 5B: Groupby Single Column, Multiple Aggregations

Often, you'll want to apply several aggregation functions to a single column within a group. You can do this by passing a list of aggregation functions to `.agg()`.

````python
state_count_stats = mock_regional_data.groupby('State').agg(min_count=('Count', 'min'), max_count=('Count', 'max'), avg_count=('Count', 'mean'))
print("--- Groupby State with Min, Max, and Avg Count ---")
display(state_count_stats)
````

#### 5C: Groupby Multiple Columns, Single Aggregation

You can also group by multiple columns to create more granular aggregations. Here, we group by both `State` and `Year` to see the total `Count` for each state in each year.

````python
state_year_sum = mock_regional_data.groupby(['State', 'Year']).agg(total_count=('Count', 'sum'))
print("--- Groupby State and Year with Total Count ---")
display(state_year_sum)
````

#### 5B: Advanced Filtering over Compiled Groups via Lambdas

After grouping, you can apply further filters to the groups themselves. The `.filter()` method, often used with a `lambda` function, allows you to keep or discard entire groups based on a condition.

````python
filtered_groups = mock_regional_data.groupby('State').filter(lambda x: x['Count'].sum() > 500)
print("\n--- Filtered Groups via Lambda Condition (> 500 total count) ---")
display(filtered_groups)
````

#### 5E: Demonstrating Outer Merge

An `outer` merge includes all rows from both DataFrames, filling `NaN` for values that don't have a match in the other DataFrame. It's useful when you want to preserve all information from both sides.

````python
merged_outer = pd.merge(left_table, right_table, on='Emp_ID', how='outer')
print("\n--- Outer Relational Join Merged Data Matrix ---")
display(merged_outer)
print("Difference: Includes all rows from both tables, with NaN for unmatched entries.")
````

#### 5F: Demonstrating Left Merge

A `left` merge includes all rows from the *left* DataFrame and matching rows from the *right* DataFrame. If there's no match on the right, it fills with `NaN`. This is useful for adding information to your primary DataFrame without losing any of its original records.

````python
merged_left = pd.merge(left_table, right_table, on='Emp_ID', how='left')
print("\n--- Left Relational Join Merged Data Matrix ---")
display(merged_left)
print("Difference: Keeps all rows from the left table and adds matching data from the right.")
````

#### 5G: Demonstrating Right Merge

A `right` merge is the inverse of a left merge: it includes all rows from the *right* DataFrame and matching rows from the *left* DataFrame. Unmatched values from the left are filled with `NaN`.

````python
merged_right = pd.merge(left_table, right_table, on='Emp_ID', how='right')
print("\n--- Right Relational Join Merged Data Matrix ---")
display(merged_right)
print("Difference: Keeps all rows from the right table and adds matching data from the left.")
````

#### 5C: Inner vs Outer Relational Joins & Key Discrepancy Scans

`pd.merge()` is essential for combining DataFrames based on common columns (keys). The `how` parameter (e.g., `'inner'`, `'outer'`, `'left'`, `'right'`) dictates how rows are handled when keys don't match across DataFrames.

Here, we demonstrate an `inner` merge, which only keeps rows where the `Emp_ID` exists in both tables.

````python
merged_inner = pd.merge(left_table, right_table, on='Emp_ID', how='inner')
print("\n--- Inner Relational Join Merged Data Matrix ---")
display(merged_inner)
````

## Part 7: Regular Expressions (Regex) Text Identification
Regex expressions act as structured filter masks designed to locate complex matching character patterns across text vectors:
* `\d`: Evaluates numeric digit variations (equivalent to `[0-9]`).
* `\w`: Targets alphanumeric string characters (equivalent to `[A-Za-z0-9]`).
* `+`: Greedy quantifier representing one or more continuous character occurrences.

````python
import re

system_log = "Transaction verified by operator 405 on machine index 9821. Critical error log reference code ID: 8812."
numeric_search_pattern = r'\d+'

# 7A: re.findall() Multi-match Scanning
all_matches = re.findall(numeric_search_pattern, system_log)
print("Discovered Numeric Array Match Matches:", all_matches)
````

#### 7B: `re.search()` for First Match

`re.search()` scans through a string looking for the first location where the regular expression produces a match. If a match is found, `re.search()` returns a match object; otherwise, it returns `None`.

````python
search_match = re.search(numeric_search_pattern, system_log)
if search_match:
    print("re.search Execution (first numeric match):", search_match.group())
else:
    print("No match found with re.search.")
````

#### 7C: `re.match()` for Start-of-String Match

`re.match()` checks for a match only at the beginning of the string. If the pattern is found at the start, it returns a match object; otherwise, it returns `None`. Note that our `numeric_search_pattern` will not match at the very beginning of `system_log`.

````python
match_result = re.match(numeric_search_pattern, system_log)
if match_result:
    print("re.match Execution (starts at string beginning):", match_result.group())
else:
    print("No match found at the beginning of the string with re.match.")
````
